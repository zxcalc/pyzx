```python
import pyzx

g = pyzx.generate.cnots(16, 48, seed=42)
g.add_vertex(ty=pyzx.VertexType.X, qubit=0, row=32.40)
g.add_vertex(ty=pyzx.VertexType.X, qubit=0, row=32.80)
g.add_vertex(ty=pyzx.VertexType.X, qubit=0, row=33.20)

pyzx.draw(g, labels=True, show_scalar=True)
```
