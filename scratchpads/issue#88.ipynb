```python
import pyzx

g = pyzx.generate.cnots(16, 48, seed=42)
g.add_vertex(ty=pyzx.VertexType.X, qubit=0, row=32.40)
g.add_vertex(ty=pyzx.VertexType.X, qubit=6, row=19.40)
g.add_vertex(ty=pyzx.VertexType.X, qubit=6, row=31.00)

pyzx.draw(g, labels=True)
```
