```python
import itertools

import pyzx as zx
from pyzx.rewrite_rules.wrules_bzw import check_bialgebra_zw_forward, check_bialgebra_zw_reverse, apply_bialgebra_zw_forward
```

```python
g = zx.Graph()

i0 = g.add_vertex(qubit=0, row=0)
i1 = g.add_vertex(qubit=1, row=0)
wi0 = g.add_vertex(ty=zx.VertexType.W_INPUT, qubit=0, row=1)
wi1 = g.add_vertex(ty=zx.VertexType.W_INPUT, qubit=1, row=1)
wo0 = g.add_vertex(ty=zx.VertexType.W_OUTPUT, qubit=0, row=2)
wo1 = g.add_vertex(ty=zx.VertexType.W_OUTPUT, qubit=1, row=2)
z0 = g.add_vertex(ty=zx.VertexType.Z, qubit=0, row=3)
z1 = g.add_vertex(ty=zx.VertexType.Z, qubit=1, row=3)
o0 = g.add_vertex(qubit=0, row=4)
o1 = g.add_vertex(qubit=1, row=4)

g.add_edge((i0, wi0))
g.add_edge((i1, wi1))
g.add_edge((wi0, wo0))
g.add_edge((wi1, wo1))
for w, z in itertools.product([wo0, wo1], [z0, z1]):
    g.add_edge((w, z))
g.add_edge((z0, o0))
g.add_edge((z1, o1))

print(f"BZW forward applicable : {check_bialgebra_zw_forward(g, ws=(wo0, wo1), zs=(z0, z1))}")

zx.draw(g, labels=True)

apply_bialgebra_zw_forward(g, ws=(wo0, wo1), zs=(z0, z1))

zx.draw(g, labels=True)
```

```python
g = zx.Graph()

i0 = g.add_vertex(qubit=0, row=0)
i1 = g.add_vertex(qubit=1, row=0)
z = g.add_vertex(ty=zx.VertexType.Z, qubit=0.5, row=1)
wi = g.add_vertex(ty=zx.VertexType.W_INPUT, qubit=0.5, row=2)
wo = g.add_vertex(ty=zx.VertexType.W_OUTPUT, qubit=0.5, row=3)
o0 = g.add_vertex(qubit=0, row=4)
o1 = g.add_vertex(qubit=1, row=4)

g.add_edge((i0, z))
g.add_edge((i1, z))
g.add_edge((z, wi))
g.add_edge((wi, wo))
g.add_edge((wo, o0))
g.add_edge((wo, o1))

print(f"BZW reverse applicable : {check_bialgebra_zw_reverse(g, w=wo, z=z)}")

zx.draw(g, labels=True)
```
