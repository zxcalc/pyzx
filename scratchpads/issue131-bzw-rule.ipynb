```python
import itertools

import pyzx as zx
from pyzx import EdgeType
from pyzx.rewrite_rules.wrules_bzw import check_bialgebra_zw_reverse, check_bialgebra_zw_forward, apply_bialgebra_zw_reverse
```

```python
g = zx.Graph()

wos = []
zs = []

for q in range(3):
    i = g.add_vertex(qubit=q, row=0)
    wi = g.add_vertex(ty=zx.VertexType.W_INPUT, qubit=q, row=1)
    wo = g.add_vertex(ty=zx.VertexType.W_OUTPUT, qubit=q, row=2)
    z = g.add_vertex(ty=zx.VertexType.Z, qubit=q, row=3)
    o = g.add_vertex(qubit=q, row=4)

    wos.append(wo)
    zs.append(z)

    g.add_edge((i, wi))
    g.add_edge((wi, wo), EdgeType.W_IO)
    g.add_edge((z, o))

for wo, z in itertools.product(wos, zs):
    g.add_edge((wo, z))

print(f"BZW reverse applicable : {check_bialgebra_zw_reverse(g, wos, zs)}")

zx.draw(g, labels=True)

apply_bialgebra_zw_reverse(g, wos, zs)

zx.draw(g, labels=True)
```

```python
g = zx.Graph()

qubits = 3
average_qubit = (qubits-1) / 2
z = g.add_vertex(ty=zx.VertexType.Z, qubit=average_qubit, row=1)
wi = g.add_vertex(ty=zx.VertexType.W_INPUT, qubit=average_qubit, row=2)
wo = g.add_vertex(ty=zx.VertexType.W_OUTPUT, qubit=average_qubit, row=3)

g.add_edge((z, wi))
g.add_edge((wi, wo), EdgeType.W_IO)

for q in range(qubits):
    i = g.add_vertex(qubit=q, row=0)
    g.add_edge((i, z))
    o = g.add_vertex(qubit=q, row=4)
    g.add_edge((wo, o))

print(f"BZW forward applicable : {check_bialgebra_zw_forward(g, w=wo, z=z)}")

zx.draw(g, labels=True)
```
